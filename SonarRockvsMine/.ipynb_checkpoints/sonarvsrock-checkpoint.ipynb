# Sonar VS Rock

```python
# Import Libraries
import numpy as np # For Math
import pandas as pd # For Data Frame
import matplotlib.pyplot as plt # For Graph
import seaborn as sns # For Advence Graph

from sklearn.model_selection import train_test_split # For Train & Test Split
from sklearn.linear_model import LogisticRegression # For Logistic Regression
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score, mean_absolute_error # For Accuracy Score
from sklearn.preprocessing import LabelEncoder, StandardScaler # For Labelling & Scalling

import warnings # For Ingnore Unneeded Warnings
warnings.filterwarnings('ignore')
```

# Data PreProcessing

```python
# Load Data
df = pd.read_csv('sonar data.csv', header=None)
df
```

# Model

```python
# Split
target = 60
features = [col for col in df.columns if col != target]
x = df[features]
y = df[target]
x_train, x_test, y_train, y_test = train_test_split(x, y, train_size = .8, random_state=23)
```

```python
# Labelling
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
# Scalling
scaler = StandardScaler()
x_train_scale = scaler.fit_transform(x_train)
x_test_scale = scaler.transform(x_test)
```

```python
# Define The Model
model = LogisticRegression(max_iter=10000, random_state=0)
model.fit(x_train, y_train)
```

```python
# Predict Test
y_pred = model.predict(x_test)
y_pred
```

```python
# Evaluate Model Performence
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
acc = accuracy_score(y_test, y_pred) * 100
```

```python
# Print The Result
print(f'Mean Absolute Error: {mae:.2f}%')
print(f'Mean Squared Error: {mse:.2f}%')
print(f'R² Score: {r2:.2f}%')
print(f"Rock VS Mine accuracy Score: {acc:.2f}%")
```

# Conclusion

# The logistic regression model was successfully trained to classify sonar signals as either rocks or mines. The model's performance on the test data showed:

# 1. Mean Absolute Error (MAE): Indicates the average magnitude of errors in predictions.

# 2. Mean Squared Error (MSE): Reflects the average squared difference between predicted and actual values.

# 3. R² Score: Shows how well the model explains the variance in the target variable.

# 4. Accuracy Score: Indicates the overall classification accuracy for distinguishing rocks from mines.

# The reported accuracy suggests that the model performs well on the dataset and can be considered reliable for basic classification between sonar signals reflected from rocks versus mines.
